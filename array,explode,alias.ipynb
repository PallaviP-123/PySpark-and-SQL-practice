```python
data = [('Pallavi', 'Dancing,singing'), ('Rajesh', 'Singing,acting'), ('Priya', 'Dancing,sports')]
columns = ['Name', 'Hobbies']

df = spark.createDataFrame(data, columns)
df.show()
```

```python
#Let's split the 'Hobbies' column into multiple columns, one for each hobby.
from pyspark.sql.functions import split

df1=df.select(df.Name,split(df.Hobbies,',').alias("Hobbies"))
display(df1)
```

```python
from pyspark.sql.functions import explode

df2 = df1.withColumn("Hobbies",explode(df1.Hobbies))
display(df2)
```
