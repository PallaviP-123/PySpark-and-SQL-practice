###Requirement:
####1. Answer the below question using the dev.spark_db.sf_fire_calls
What are top 3 zip codes that accounted for most calls?

```python
%sql

-- Answer using a SQL query
select CallType, Zipcode, count(*) as count
from dev.spark_db.sf_fire_calls
where CallType is not null
group by CallType, Zipcode
order by count desc
limit 3
```

#### 2. Answer the above question using the Spark DataFrame API

2.1 Create a DataFrame reading data from the table

```python
# Read data and create a dataframe

fire_df = spark.read.table("dev.spark_db.sf_fire_calls")
```

2.2 Apply Transformations

```python
# Apply necessory transformations
df_1 = fire_df.select("CallType", "Zipcode")
df_2 = df_1.where("CallType is not null")
df_3 = df_2.groupBy("CallType", "Zipcode").count()
df_4 = df_3.orderBy("count", ascending=False)
df_5 = df_4.limit(3)
```

2.3 Apply an Action

1. Optimize the query
2. Execute the optimized plan

####3. Spark Dataframe API concepts
1. Spark creates optimized query plan
2. Dataframes are immutable
3. Every Transformation returns a dataframe
2. Dataframe offers composable API

1. How to see the optimized query plan
    1. Use explain method
    2. Check the query runtime profile

```python
df_5.explain(mode="extended")
```

2. Dataframes are immutable\
You can see or use any intermediate dataframe

```python
fire_df.display()
```

3. Every Transformation returns a dataframe
4. Dataframe offers composable API\
  Example of composable API

```python
result_df = (
    fire_df.select("CallType", "Zipcode")
            .where("CallType is not null")
            .groupBy("CallType", "Zipcode")
            .count()
            .orderBy("count", ascending=False)
            .limit(3)
)

result_df.display()
```
