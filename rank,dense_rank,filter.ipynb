Department wise the nth highest salary of employees

```python
data=[("1","A","1000","IT"),("2","B","2000","HR"),("3","C","3000","IT"),("4","D","4000","HR"),("5","E","5000","IT"),("6","F","6000","HR"),("7","G","7000","Sales"),("8","H","9000","Sales"),("9","I","9000","Sales"),("10","J","10000","IT")]
columns=[("Empid"),("EmpName"),("Salary"),("DeptName")]
df= spark.createDataFrame(data,columns)
display(df)

df.createOrReplaceTempView("df")
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import *

df3 = df.select('*',dense_rank().over(Window.partitionBy(df.DeptName).orderBy(df.Salary.desc())).alias('rank'))
display(df3)
```

```python
df1 = spark.sql("""select *,rank() OVER(PARTITION BY DeptName ORDER BY Salary DESC) AS Rank from df""")
display(df1)
df1.createOrReplaceTempView("df1")
```

```python
display(df3.filter(df3.rank==1))
```

```python
df2 = spark.sql("""Select * from df1 where Rank=1""")
display(df2)
```
