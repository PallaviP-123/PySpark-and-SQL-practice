```python
#Employee salary info
data1=[(100,"Raj",None,1,"01-04-23",50000),
       (200,"Joanne",100,1,"01-04-23",4000),
       (200,"Joanne",100,1,"13-04-23",4500),
       (200,"Joanne",100,1,"14-04-23",4020)]
schema1=["EmpId","EmpName","ManagerId","DeptId","Date","Salary"]
df_salary=spark.createDataFrame(data1,schema1)
display(df_salary)

#Department datafram
data2=[(1,"IT"),
       (2,"HR"),]
schema2=["DeptId","DeptName"]
df_dept=spark.createDataFrame(data2,schema2)
display(df_dept)
```

```python
from pyspark.sql.functions import *

df=df_salary.withColumn('NewSalDt',to_date('Date','dd-MM-yy'))
display(df)
```

```python
df1=df.join(df_dept,['DeptId'])
display(df1)
df2=df1.alias('a').join(df1.alias('b'),col('a.ManagerId')==col('b.EmpId'),'left').select(
    col('a.DeptName'),
    col('b.EmpName').alias('ManagerName'),
    col('a.EmpName'),
    col('a.Salary'),
    col('a.NewSalDt')
)
display(df2)
```

```python
from pyspark.sql.functions import *
df3=df2.groupBy('DeptName','ManagerName','EmpName',year('NewSalDt').alias('Year'),month('NewSalDt').alias('Month')).sum('Salary')

display(df3)
```
